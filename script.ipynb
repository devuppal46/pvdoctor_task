# GHI Data Merger
Merges all CSV files from the GHI folder into a single CSV and JSON file.

```python
import os
import pandas as pd

base_path = 'GHI'

data = []

for folder in sorted(os.listdir(base_path)):

    folder_path = os.path.join(base_path,folder)

    if os.path.isdir(folder_path):
        for file in sorted(os.listdir(folder_path)):

            if file.endswith('.csv'):
                file_path = os.path.join(folder_path,file)
                df = pd.read_csv(file_path)
                data.append(df)

final_df = pd.concat(data, ignore_index=True)
final_df.to_csv('ghi_data.csv', index=False)
final_df.to_json("ghi_data.json", orient = "records", indent=2)
```

## Q1 A) What assumptions did you make about the data?

1. Each data point is continuous and there is no break in the data.
2. CSV column names are always consistent i.e Date and GHI, across all files.
3. No missing, null, or corrupt data in any row.
4. Duplicate records are'nt supported.

## Q1 B) What would break your solution?

1. if columns are different , it will create extra columns instead of merging.
2. Duplicate dates would create redundant data records.
3. If the file is not csv or empty csv files will cause problems in concatenation.
4. Large Dataset can make it crash ,because we are loading everything at once(instead of chunking).
